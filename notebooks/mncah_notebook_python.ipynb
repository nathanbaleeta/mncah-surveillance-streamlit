```python
import pandas as pd
import duckdb
```

```python
df = pd.read_excel("sample_data/GLOBAL_DATAFLOW_2000-2026.xlsx", sheet_name='Unicef data')
```

```python
# Convert to lowercase and replace spaces with dashes
df.columns = df.columns.str.lower().str.replace(' ', '_')

df.head()
```

```python
# Convert year columns to numeric and replace with zero for nulls
columns_to_cast = list(range(2000, 2025))
year_cols = [str(x) for x in columns_to_cast]

df[year_cols] = df[year_cols].apply(pd.to_numeric, errors='coerce').fillna(0).astype(int)
```

```python
df.head()
```

```python
# Identify the columns to keep as identifiers
id_vars = [col for col in df.columns if col not in year_cols]

# Melt the dataframe
df_melted = pd.melt(
    df,
    id_vars=id_vars, #  The non-year columns  serve as the identifier variables
    value_vars=year_cols,
    var_name='year',
    value_name='mncah_total'
)

# Display the melted dataframe
df_melted.head()
```

```python
# Save the DataFrame to a CSV file
df_melted.to_csv('sample_data/mncah_data_2000_2025.csv', index=False)
```

```python
df_summary = duckdb.query("""
                              SELECT geographic_area, SUM(mncah_total)
                              FROM df_melted
                              GROUP BY geographic_area, mncah_total
                              ORDER BY mncah_total
                              DESC LIMIT 10
                              """).to_df()
df_summary
```
